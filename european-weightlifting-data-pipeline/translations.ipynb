```python
# Create the translation dictionary
column_translation = {
    'Eventos': 'Events',
    'Oro': 'Gold',
    'Plata': 'Silver',
    'Bronce': 'Bronze',
    'Año': 'Year',
    'Genero': 'Gender'
}

df_2019 = df_2019.rename(columns=column_translation)
df_2020 = df_2020.rename(columns=column_translation)
```

```python
traduccion = {
    'Masculino' : 'Male',
    'Femenino' : 'Female'
}

df_2019['Gender'] = df_2019['Gender'].map(traduccion)
df_2020['Gender'] = df_2020['Gender'].map(traduccion)
```

```python
traduccion_paises = {
    'Italia': 'Italy',
    'Bulgaria': 'Bulgaria',
    'Turquía': 'Turkey',
    'Bielorrusia': 'Belarus',
    'Rumania': 'Romania',
    'Georgia': 'Georgia',
    'Armenia': 'Armenia',
    'Rusia': 'Russia',
    'Francia': 'France',
    'Ucrania': 'Ukraine',
    'Polonia': 'Poland',
    'Letonia': 'Latvia',
    'España': 'Spain',
    'Moldavia': 'Moldova',
    'Alemania': 'Germany',
    'Bélgica': 'Belgium',
    'Reino Unido': 'United Kingdom',
    'Azerbaiyán': 'Azerbaijan',
    'Eslovaquia': 'Slovakia',
    'Grecia': 'Greece',
    'Finlandia': 'Finland',
    'Austria': 'Austria',
    'República Checa': 'Czech Republic',
    'Israel': 'Israel',
    'Albania': 'Albania',
    'Suecia' : 'Sweden'
}
```

```python
def traducir_paises_en_celda(texto, diccionario):
    if pd.isna(texto):
        return texto
    
    # Convertimos a string por seguridad
    texto_str = str(texto)
    
    # Buscamos cada país en español dentro del texto y lo reemplazamos por su traducción
    for pais_es, pais_en in diccionario.items():
        if pais_es in texto_str:
            # Reemplazamos el país en español por el de inglés
            texto_str = texto_str.replace(pais_es, pais_en)
            
    # Opcional: Limpiamos los molestos códigos HTML como '&nbsp;' si quedan residuos
    texto_str = texto_str.replace('&nbsp;', ' ').replace('\xa0', ' ')
    
    return texto_str

# Aplicamos la función a las columnas de medallas en ambos DataFrames
columnas_medallas = ['Gold', 'Silver', 'Bronze']

for col in columnas_medallas:
    df_2019[col] = df_2019[col].apply(lambda x: traducir_paises_en_celda(x, traduccion_paises))
    df_2020[col] = df_2020[col].apply(lambda x: traducir_paises_en_celda(x, traduccion_paises))
```
